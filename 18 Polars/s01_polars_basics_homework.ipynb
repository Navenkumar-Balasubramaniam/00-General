## Exercise 3: Select Columns (15 points)

Create a new DataFrame with only the following columns:
- `track_name`

## Exercise 4: Create New Columns (20 points)

Using `with_columns()`, add the following columns to the `tracks` DataFrame:

1. `duration_min`: Duration in minutes (convert `duration_ms` to minutes by dividing by 60000)
2. `streams_millions`: Streams in millions (divide `streams` by 1,000,000)
3. `track_name_upper`: Track name in uppercase

Store the result in `tracks_enhanced`.

```python
# Your code here
tracks_enhanced = tracks.with_columns(
    (pl.col("duration_ms")/60000).alias("duration_min"),
     (pl.col("streams")/1000000).alias("streams_millions"),
    (pl.col("track_name").str.to_uppercase()).alias("track_name_upper")
)

# Show relevant columns
tracks_enhanced.select(
    "track_name", "duration_ms", "duration_min",
    "streams", "streams_millions", "track_name_upper"
).head()
```

## Exercise 5: Aggregations (20 points)

Calculate the following statistics for the `tracks` DataFrame in a single `select()` statement:

1. Average popularity (`avg_popularity`)
2. Maximum streams (`max_streams`)
3. Minimum tempo (`min_tempo`)
4. Number of unique artists (`unique_artists`)
5. Total number of tracks (`total_tracks`)

```python
# Your code here
stats = tracks.select(
    pl.col("popularity").mean().alias("avg_popularity"),
    pl.col("streams").max().alias("max_streams"),
    pl.col("tempo").min().alias("min_tempo"),
    pl.col("artist").n_unique().alias("unique_artists"),
    pl.col("track_name").count().alias("total_tracks")
)
stats
```

## Exercise 7: Save to Parquet (15 points)

1. Save the `tracks_enhanced` DataFrame to a Parquet file called `tracks_enhanced.parquet`
2. Read it back into a new DataFrame called `tracks_from_parquet`
3. Verify the shape matches the original

```python
# 7.1 Save to Parquet
tracks_enhanced.write_parquet("tracks_enhanced.parquet")
```

```python
# 7.2 Read back from Parquet
tracks_from_parquet = pl.read_parquet("tracks_enhanced.parquet")
```

```python
# 7.3 Verify shapes match
print(f"Original shape: {tracks_enhanced.shape}")
print(f"From Parquet shape: {tracks_from_parquet.shape}")
print(f"Shapes match: {tracks_enhanced.shape == tracks_from_parquet.shape}")
```

## Bonus Exercise: Complex Expression (10 bonus points)

Create a new column called `energy_category` based on the `energy` value:
- If energy >= 0.8: "High Energy"
- If energy >= 0.5: "Medium Energy"
- Otherwise: "Low Energy"

Hint: You'll need to use `pl.when().then().otherwise()` which we'll cover in detail in Session 2, but try to figure it out!

```python
# Bonus: Your code here
tracks_with_category = tracks.with_columns(
    (pl.when(pl.col("energy") >= 0.8).then(pl.lit("High Energy"))
    .when(pl.col("energy") >= 0.5).then(pl.lit("Medium Energy")).otherwise(pl.lit("Low Energy"))).alias("energy_category")
)

# Show sample results
tracks_with_category.select("track_name", "energy", "energy_category").head(10)
```
