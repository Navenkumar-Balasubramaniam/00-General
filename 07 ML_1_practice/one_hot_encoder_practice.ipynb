```python
url = "https://raw.githubusercontent.com/Navenkumar-Balasubramaniam/00-General/refs/heads/main/07%20ML_1_practice/customer_membership_payment_spend_augmented.csv"
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.read_csv(url)
```

```python
df.head()
```

```python
categorical_list = ["Membership_Type","Payment_Method","Region","Signup_Channel","Spend_Bucket","Account_Notes"]
ohe = OneHotEncoder(drop = "first",sparse_output = False)
cat_arr = ohe.fit_transform(df.loc[:,categorical_list])
```

```python
print(cat_arr)
```

```python
ohe_columns = ohe.get_feature_names_out(categorical_list)
```

```python
cat_df = pd.DataFrame(
    cat_arr,
    columns = ohe_columns,
    index = df.index
)
```

```python
df_encoded = pd.concat(
    [df,cat_df],
    axis = 1
)
```

```python
df_encoded.head()
```

```python
df_encoded["Monthly_spend_avg"] = df.groupby("Payment_Method")["Monthly_Spend"].transform("mean")
```
